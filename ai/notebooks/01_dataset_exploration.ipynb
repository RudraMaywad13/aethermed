```python
import io
import redivis
from PIL import Image
from IPython.display import display  # <-- Import the inline display tool

def show_image(file):
    table = redivis.organization("aimi").dataset("chexpert_plus").table("png_train")
    image_file = table.file(f"{file}")
    image_bytes = image_file.read()

    with io.BytesIO(image_bytes) as f:
      img = Image.open(f)
      img.load()


      display(img)
show_image('s6cj-f5e10m11h.knrKnfELpsfrDuUpsQsEmg')
```
